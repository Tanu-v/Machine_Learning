#Simple Linear Regression performed

```python
import matplotlib.pyplot as plt
import pandas as pd
dataset = pd.read_excel('Book1.xlsx')
X = dataset.iloc[:, 0:1].values
y = dataset.iloc[:, -1].values
plt.scatter(X, y, color='red')
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 5)
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg.fit(X_train, y_train  )
y_pred = reg.predict(X_test)
from sklearn.metrics import r2_score
r_square=r2_score(y_test,y_pred)
reg.predict([[2020],[2021]])
plt.scatter(X_test, y_test, color = 'red')
plt.plot(X_train, reg.predict(X_train), color = 'blue')
plt.title('PER CAPITA INCOME OF CANADA')
plt.xlabel('Year')
plt.ylabel('Per Capita Income')
plt.show()
```

```python
from sklearn.metrics import r2_score
y=[1,2,3,4,5]
y1=[0.8,2.5,3,4.8,4.4]
r_square=r2_score(y,y1)
print("Co-efficient of determination:- ",r_square)
print("Accuracy:",r_square*100)
print("Error:",100-r_square*100)
```
