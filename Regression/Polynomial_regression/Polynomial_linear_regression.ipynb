```python
import pandas as pd
import matplotlib.pyplot as plt
spd = pd.read_csv('/content/Salary_position.csv')
print(spd)
```

```python
x = spd.iloc[:,1:2].values
x
```

```python
y = spd.iloc[:,2].values
y
```

```python
plt.scatter(x,y,color='red')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
rgr = LinearRegression()
rgr.fit(x,y)
```

```python
y_pred = rgr.predict(x)
y_pred
```

```python
plt.scatter(x,y,color='red')
plt.plot(x,y_pred,color='blue')
plt.show()
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly_reg = PolynomialFeatures(degree=4)
x_1 = poly_reg.fit_transform(x)
print(x_1)
```

```python
from sklearn.linear_model import LinearRegression
lin_reg = LinearRegression()
lin_reg.fit(x_1,y)
```

```python
y_pred = lin_reg.predict(x_1)
y_pred
```

```python
plt.scatter(x,y,color='red')
plt.plot(x,y_pred,color='blue')
plt.show()
```

```python
for i in range(1,9):
  poly_reg = PolynomialFeatures(degree=i)
  x_1 = poly_reg.fit_transform(x)
  lin_reg = LinearRegression()
  lin_reg.fit(x_1,y)
  plt.scatter(x,y,color='red')
  plt.plot(x,lin_reg.predict(x_1),color='blue')
  plt.show()
  print('-------------------')
```

```python
poly_reg = PolynomialFeatures(degree=7)
```

```python
x_1 = poly_reg.fit_transform(x)
```

```python
lin_reg = LinearRegression()
lin_reg.fit(x_1,y)
```

```python
from sklearn.metrics import r2_score
r2_score(y,lin_reg.predict(x_1))
```

```python
lin_reg.score(x_1,y)
```
