```python
import pandas as pd
ds = pd.read_csv('/content/ml intial data.csv')
print(ds)
```

```python
sns.scatterplot(data=ds, x='Area', y='Price')
```

```python
from sklearn.linear_model import LinearRegression
Reg = LinearRegression()
```

```python
Reg.fit(ds[['Area']], ds.Price)
```

```python
Reg.predict([[3500]])
```

```python
Reg.coef_
```

```python
Reg.intercept_
```

```python
from sklearn.metrics import r2_score
y_original = ds.Price
y_predicted = Reg.predict(ds[['Area']])
R_square = r2_score(y_original, y_predicted)
print('R squared value',R_square)
```
