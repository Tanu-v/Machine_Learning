##load_digits dataset

```python
import pandas as pd
from sklearn.datasets import load_digits
digits = load_digits()
```

```python
dir(digits)
```

```python
import matplotlib.pyplot as plt
plt.gray()
for i in range(10):
    plt.matshow(digits.images[i])
```

```python
df = pd.DataFrame(digits.data)
df.head()
```

```python
df['target'] = digits.target
df.head()
```

```python
x = df.drop(['target'], axis='columns')
y = digits.target
print(x)
```

```python
print(y)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier()
model.fit(x_train, y_train)
model.score(x_test,y_test)
```

```python
model5 = RandomForestClassifier(n_estimators=50)
model5.fit(x_train, y_train)
```

```python
model5.score(x_test,y_test)
```

```python
model5.predict([digits.data[12]])
```

##Salary data

---



---



```python
dfs = pd.read_csv('/content/Salary_Data.csv')
dfs.head()
```

```python
dfs.info()
```

```python
dfs.dropna(inplace=True)
dfs
```

```python
x = dfs[['Years of Experience']]
print(x.head())
```

```python
y = dfs['Salary'].values
y
```

```python
from sklearn.ensemble import RandomForestRegressor
mo = RandomForestRegressor(n_estimators=100 , random_state=0)
mo.fit(x,y)
```

```python
mo.score(x,y)
```

```python
mo.predict([[4.2]])
```

```python
plt.scatter(x,y,color='blue')
```

```python
plt.plot(x,mo.predict(x),color='red')
```

```python
plt.scatter(x,y,color='blue')
plt.plot(x,mo.predict(x),color='red')
plt.title('Random Forest Regression')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```
