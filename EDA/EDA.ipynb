```python
def Category_to_num(x):
  if x=='Labels':
    return 1
  if x=='Furnishings':
    return 2
  if x=='Art':
    return 3
  if x=='Phones':
    return 4
  if x=='Binders':
    return 5
  if x=='Appliances':
    return 6
  if x=='Tables':
    return 7
  if x=='Storage':
    return 8
  if x=='Accessories':
    return 9
  if x=='Paper':
    return 10
  if x=='Chairs':
    return 11
  if x=='Fasteners':
    return 12
  if x=='Supplies':
    return 13
  if x=='Envelopes':
    return 14
  if x=='Copiers':
    return 15
  if x=='Bookcases':
    return 16
  if x=='Machines':
    return 17
```

```python
pf_m['Category']=pf_m['Category'].apply(Category_to_num)
```
