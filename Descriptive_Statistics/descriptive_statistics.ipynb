```python
import pandas as pd
df = pd.read_csv('/content/processed_aqi.csv')
df.head()
```

#1.Mean

```python
mean = df['AQI'].mean()
print(mean)
```

#2.Median

```python
median = df['AQI'].median()
print(median)
```

#3.Mode

```python
from scipy import stats
mode = stats.mode(df['AQI'])
print(mode)
```

#4.Standard deviation

```python
std_dev = df['AQI'].std()
print(std_dev)
```

#5.Variance

```python
variance=df['AQI'].var()
print(variance)
```

#6.Minimum

```python
min_value=df['AQI'].min()
print(min_value)
```

#7.Maximum

```python
max_value=df['AQI'].max()
print(max_value)
```

#8.Percentile

```python
percentile = df['AQI'].quantile(0.9)
print(percentile)
```

#9.Kurtosis

```python
from scipy.stats import kurtosis
kurtosis_value = kurtosis(df['AQI'])
print(kurtosis_value)
```

#10.Skewness

```python
from scipy.stats import skew
skewness_value = skew(df['AQI'])
print(skewness_value)
```

#11.Correlation

```python
correlation=df['AQI'].corr(df['PM2.5'])
print(correlation)
```
