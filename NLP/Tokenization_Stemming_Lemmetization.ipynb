```python
from nltk import word_tokenize,sent_tokenize
tok = "I want eat dosa."
print(word_tokenize(tok))
```

```python
sp = "Taniya is nagging me .\n she wants to go out ."
print(word_tokenize(sp))
```

```python
st = "Lets go out . \n being in home is boring ."
print(sent_tokenize(st))
```

```python
para = "Natural Language Processing (NLP) is a branch of artificial intelligence (AI) that enables computers to understand, interpret, manipulate, and generate human language in a meaningful and useful way. \n By bridging the communication gap between humans and machines, NLP allows devices to process both written text and spoken speech using a combination of computational linguistics, machine learning, and deep learning."
print(word_tokenize(para))
print(sent_tokenize(para))
```

```python
import nltk
n = nltk.word_tokenize("Ram is a good boy.")
s = nltk.sent_tokenize("Ram is a good boy.")
print(n)
print(s)
```

```python
from nltk import word_tokenize
w = word_tokenize("I am a good boy.")
p = sent_tokenize("I am a good boy.")
print(w)
print(p)
```

```python
from nltk import *
s1 = word_tokenize("I am a good boy.")
s2 = sent_tokenize("I am a good boy.")
print(s1)
print(s2)
```

```python
from nltk import word_tokenize as wt
t = wt("I am a good boy.")
print(t)
```

```python
from nltk.tokenize import TreebankWordTokenizer as a
q = a().tokenize("I am a good boy.")
print(q)

```

```python
from nltk.tokenize import TreebankWordTokenizer as a
q = a()
z = q.tokenize("I am a good boy.")
z
```

```python
import nltk
i = nltk.tokenize.TreebankWordTokenizer()
i.tokenize("you can visit my website www.ramza.com.")
```

```python
i1 = nltk.word_tokenize("you can visit my website www.ramza.com.")
print(i1)
```

```python
i2 = nltk.word_tokenize("I don't know.")
i3 = nltk.word_tokenize("don't")  #sub word tokenize
i4 = nltk.tokenize.TreebankWordTokenizer()
i5 = i4.tokenize("can't")
i6 = nltk.word_tokenize("can't")
print(i6)
print(i2)
print(i3)
print(i5)
```

```python
from nltk.tokenize import WordPunctTokenizer as wp
r = wp()
pt1 = r.tokenize("can't")
pt2 = r.tokenize("I don't know.")
pt3 = r.tokenize("I want to eat rice\\roti.")
pt4 = r.tokenize("I want to go to website www.tv.com.")
print(pt1)
print(pt2)
print(pt3)
print(pt4)
```

```python
from nltk.tokenize import PunktSentenceTokenizer as ps
f = ps()
f1 = f.tokenize("Flowers, also known as blossoms and blooms, are the reproductive structures of flowering plants. Typically, they are structured in four circular levels around the end of a stalk. These include: sepals, which are modified leaves that support the flower; petals, often designed to attract pollinators; male stamens, where pollen is presented; and female gynoecia, where pollen is received and its movement is facilitated to the egg. When flowers are arranged in a group, they are known collectively as an inflorescence.")
print(f1)
```

```python
from nltk.tokenize import PunktSentenceTokenizer
from nltk.corpus import webtext
fi = webtext.raw("/content/corpus_eg.txt")
pst = PunktSentenceTokenizer()
tx0 = pst.tokenize(fi)
print(tx0)
sm = PunktSentenceTokenizer(fi)
tx = sm.tokenize(fi)
print(tx)
```

with corpus

```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
wn = WordNetLemmatizer()
sen = ["was"]
a = [wn.lemmatize(sen[0],pos=wordnet.VERB)]
print(a)
```

```python
sen = ["leaves"]
a = [wn.lemmatize(sen[0],pos=wordnet.VERB)]
print(a)
```

```python
sen = ["eating"]
a = [wn.lemmatize(sen[0],pos=wordnet.VERB)]
print(a)
```

```python
sen = ["studies"]
a = [wn.lemmatize(sen[0],pos=wordnet.VERB)]
print(a)
```

```python
a = [wn.lemmatize("ponies",pos=wordnet.NOUN)]
print(a)
```
