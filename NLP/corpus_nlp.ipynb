```python
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
cs = set(stopwords.words("english"))
cs1 = ["I","am","Doing","it","right","Now"]
[word for word in cs1 if word not in cs]
```

```python
cs2 = set(stopwords.words("arabic"))
cs3 = ["I","am","Doing","it","right","Now"]
[word for word in cs3 if word not in cs2]
```

```python
cs4 = set(stopwords.words("english"))
cs5 = ["I","am","a","very","brave","girl","and","a","performer","and","a","writter","with","husteling","profession","."]
[word for word in cs5 if word not in cs4]
```
