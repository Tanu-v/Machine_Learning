#Parts of speech



```python
import nltk
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk import word_tokenize
w = "Let's do it today and not make any mistake for further corrections."
print(nltk.pos_tag(word_tokenize(w)))
```

```python
w1 = "kertoenova"
print(nltk.pos_tag(word_tokenize(w1)))
```

#default tagger

```python
from nltk.tag import DefaultTagger
dt = DefaultTagger('VB')
print(dt.tag(word_tokenize(w1)))
```

```python
dt1 = DefaultTagger('DT')
dt1.tag(['I','am','rude','girl','.'])
```

#Untagging


```python
from nltk.tag import untag
o = untag([('I', 'PRP'), ('am', 'VB'), ('rude', 'JJ'), ('girl', 'NN'), ('.', '.')])
o
```

```python
nltk.pos_tag(o) #tagging after untagging and the input is already tokenized
```

#Accuracy checking through treebank


```python
import nltk
nltk.download('treebank')
```

```python
from nltk.tag import DefaultTagger
from nltk.corpus import treebank
d = DefaultTagger('VBG')
t = treebank.tagged_sents()[1000:]
d.evaluate(t)
```

#Unigram tagger

```python
from nltk.tag import UnigramTagger
from nltk.corpus import treebank
u = treebank.tagged_sents()[:2500]
u1 = UnigramTagger(u) #here we are allocating training data
u2 = treebank.tagged_sents()[1500:]
u1.evaluate(u2)
```

```python
a = treebank.tagged_sents()[:1000]
a1 = UnigramTagger(a)
treebank.sents()[0]
```

#Parsing


```python
import nltk
sentence = [
   ("a", "DT"),
   ("clever", "JJ"),
   ("fox","NN"),
   ("was","VBP"),
   ("jumping","VBP"),
   ("over","IN"),
   ("the","DT"),
   ("wall","NN")
]
grammar = "NP:{<DT>?<JJ>*<NN>}"
Reg_parser = nltk.RegexpParser(grammar)
Reg_parser.parse(sentence)
Output = Reg_parser.parse(sentence)
Output.draw()
```
